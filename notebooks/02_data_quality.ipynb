# Data quality review

Owner-run, local-only view of activity screening flags. Outputs may contain private health data; never commit notebook outputs or share them.

```python
import polars as pl

from stridelog.db.models import Activity
from stridelog.notebook_data import read_table
```

```python
activities = read_table(Activity, ["id", "name", "local_date", "dq_flags", "hr_source"])
schema = {
    "id": pl.Int64,
    "name": pl.String,
    "local_date": pl.Date,
    "dq_flags": pl.List(pl.String),
    "hr_source": pl.String,
}
pl.DataFrame(activities, schema=schema).sort("local_date", descending=True)
```

Flags are warnings, not diagnoses or automatic exclusions. Review source data locally; this notebook does not infer a cause.
