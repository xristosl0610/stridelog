# Fitness and fatigue

Owner-run, local-only view of daily derived load. Outputs may contain private health data; never commit or share them.

```python
import plotly.graph_objects as go
import polars as pl

from stridelog.db.models import DailyAnalytics
from stridelog.notebook_data import read_table
```

```python
daily = read_table(DailyAnalytics, ["local_date", "metrics"])
rows = [{"local_date": row["local_date"], **row["metrics"]} for row in daily]
if not rows:
    raise ValueError("No daily analytics are available; run stridelog analyze first.")
series = pl.DataFrame(rows).sort("local_date")
go.Figure(
    [
        go.Scatter(x=series["local_date"].to_list(), y=series[key].to_list(), name=key)
        for key in ("ctl", "atl", "tsb")
    ]
).update_layout(title="Training-load series", xaxis_title="Local date")
```

```python
go.Figure(
    [
        go.Scatter(x=series["local_date"].to_list(), y=series[key].to_list(), name=key)
        for key in ("acute_chronic_workload_ratio", "weekly_ramp_rate_percent")
    ]
).update_layout(title="Load ratio and weekly ramp rate", xaxis_title="Local date")
```
